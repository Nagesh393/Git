```python
def save_note():
    note=input("Type a note to save")
    with open("notes.txt", "w") as f:
        f.write(note)
    print("notes saved to notes.txt")
def read_file_by_name():
    file_name=input("Enter the file name to read")
    try:
        with open(file_name, 'r') as f:
            print(f.read())
    except:
        print("File not found")
def repeat_note():
    note= input("Type a note")
    try:
        times=int((input("How many time do you want to print")))
        with open("repeated.txt", 'w') as f:
            for x in range(times):
                f.write(note+"\n")
        print("saved to repeated.txt")
    except:
        print("Enter a valid number")
save_note()
read_file_by_name()
repeat_note()
read_file_by_name()



```

```python
import pandas as pd
def check_result(marks):
    if marks >= 40:
        return "Pass"
    else:
        return "Fail"
data={
    "name":["Nagesh", "Suresh", "Jay", "Dinesh", "RAJ"],
    "marks" : [99, 95, 10, 93, 90]
}
df=pd.DataFrame(data)
topscore=df['marks'].max()

df["passed"]= df['marks']>50

df["result"] = df["marks"].apply(check_result)

print(df);

pass_count = df["passed"].sum()

print(pass_count)
```
